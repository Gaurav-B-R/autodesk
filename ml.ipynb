#### Geographical Region (geo_region)

```python
# Analysis of geo_region column
total_records = len(df)
unique_regions = df['geo_region'].nunique()

# Count numeric regions
numeric_mask = df['geo_region'].str.isnumeric()
numeric_records = numeric_mask.sum()
numeric_percentage = (numeric_records / total_records) * 100

print("=== Geo Region Analysis ===")
print(f"Total unique regions: {unique_regions}")
print(f"\nNumeric regions:")
print(f"Count: {numeric_records}")
print(f"Percentage: {numeric_percentage:.2f}%")
print(f"\nMost common numeric regions (top 5):")
print(df[numeric_mask]['geo_region'].value_counts().head())

print(f"\nMost common non-numeric regions (top 5):")
print(df[~numeric_mask]['geo_region'].value_counts().head())
```

#### We can see that there are some numerical values (14.37%) present in the column. We can treat them but there is no point as we are not sure how those data are coming and does that actually have any indication. Statistically this wont be problem so we can consider them as just another categorical values and proceed.
